### Part 1: Algorithm কী?
Algorithm হলো কোনো একটি Problem সমাধান করার জন্য নির্দিষ্ট কিছু ধাপের সমষ্টি।

```python
# List দেওয়া হলো. এই List থেকে 40 খুঁজে বের করা।
numbers = [80,30,40,50,80,10,30]
print(numbers)
```

এটি করার একটি Algorithm:
1. প্রথম Element থেকে শুরু করবে।
2. প্রতিটি Element পরীক্ষা করবে।
3. যদি 40 পাওয়া যায়, তাহলে তার Index Return করবে।
4. না পাওয়া গেলে -1 Return করবে।

```python
for number in numbers:
    print(numbers[0])
```

Part 2: Time Complexity কী? <br>
Time Complexity হলো Input Size বাড়ার সঙ্গে একটি Algorithm-এর কাজের পরিমাণ কীভাবে বাড়ে তার পরিমাপ।
<br>
এখানে আমরা সাধারণত Execution Time সেকেন্ডে মাপি না। বরং Input-এর আকার অনুযায়ী কতগুলো Operation হচ্ছে তা বিশ্লেষণ করি।

1. Sequential Loops → Complexity যোগ করতে হয়।
2. Nested Loops → Complexity গুণ করতে হয়।
3. Constant Loop (range(5) বা range(10)) → Constant হিসেবে গণ্য হয়।
4. পাশাপাশি থাকা Nested Loop ও সাধারণ Loop → শেষে Dominant Term রাখতে হয়।
